```python
import pandas as pd
import glob

from faker import Faker
fake = Faker()
Faker.seed(42)

def fake_map(series, generator_func):
    unique_vals = series.dropna().unique()
    mapping = {val: generator_func() for val in unique_vals}
    return series.map(mapping)
```

```python
folder_path = r"C:\Users\gabriel.palomarez\Documents\MidYearReport\Midyear Data\Midyear_01012019_07312019 - Copy"

all_files = glob.glob(folder_path + "/**/*.xlsx", recursive=True)

df_list = []
for f in all_files:
    try:
        df = pd.read_excel(f, engine="openpyxl")
        df_list.append(df)
    except Exception as e:
        print(f"⚠️ Skipped {f} due to: {e}")

final_df = pd.concat(df_list, ignore_index=True)
```

```python
#final_df['Practice'] = final_df['Practice'].apply(lambda x: fake_name(x) if pd.notna(x) else x)
#final_df['Location'] = final_df['Location'].apply(lambda x: fake_name(x) if pd.notna(x) else x)
#final_df['Patient Group'] = final_df['Patient Group'].apply(lambda x: fake_name(x) if pd.notna(x) else x)

final_df['Patient Group'] = fake_map(final_df['Patient Group'], lambda: fake.unique.lexify(text='Group ???'))
final_df['Practice'] = fake_map(final_df['Practice'], lambda: fake.unique.company())
final_df['Location'] = fake_map(final_df['Location'], lambda: fake.unique.company() + " Medical Group")
```
